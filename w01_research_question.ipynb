## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# loading csv file
url = "https://raw.githubusercontent.com/ggirlrottingg/ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Key Data Numbers
total_pages = len(df)
declining_pct = (df["trend_direction"] == "down").mean() * 100
median_position = df["avg_position"].median()
age_position_corr = df["content_age_days"].corr(df["avg_position"], method="spearman")

# Print out clean summary numbers for your lane
print(f"Total rows analyzed: {total_pages:,}")
print(f"Percentage of pages with declining trend: {declining_pct:.1f}%")
print(f"Median organic search position: {median_position:.1f}")
print(f"Spearman correlation (Content Age vs. Avg Position): {age_position_corr:.2f}")
```
