## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

Binary Classification. We want to predict whether an user will churn (1) or stay (0) over the next 30 days so the retention team can intervene early.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.


import pandas as pd

# Create a sample DataFrame matching lane's target label
df = pd.DataFrame({
    'user_id': [101, 102, 103, 104, 105],
    'target': [0, 1, 0, 0, 1]  # 1 = Churned, 0 = Stayed
})

# Displaying class balance
print("Target Class Balance:")
print(df['target'].value_counts(normalize=True))
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Primary Metric: PR-AUC (Precision-Recall AUC) due to heavy class imbalance (only 5% churn rate). A score > 0.35 represents significant lift over random baseline.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# Calculate baseline performance (e.g., random guess or majority class performance)
baseline_acc = df['target'].value_counts(normalize=True).max()
print(f"Baseline Majority Class Accuracy: {baseline_acc:.4f}")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule like if activity < 5 then churn misses nuanced interactions, like power users whose activity drops abruptly due to a bug versus casual users who naturally log in less often. Machine learning can combine usage trends, support ticket volume, and billing history simultaneously.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# Show data complexity, e.g., overlapping distributions or correlation matrix
df.groupby('target')[['usage_score', 'support_tickets', 'tenure_days']].mean()
```
